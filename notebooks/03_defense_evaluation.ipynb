# Defense Evaluation

Compare defense effectiveness.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from rmbench.defenses import *

sns.set_style('whitegrid')
```

## Defense Comparison

```python
import json

with open('../results/example_results.json') as f:
    results = json.load(f)

defense_data = results['defense_effectiveness']
defenses = list(defense_data.keys())
reductions = [defense_data[d]['avg_reduction_in_asr'] for d in defenses]

plt.figure(figsize=(10, 6))
plt.barh(defenses, reductions, color='green')
plt.xlabel('ASR Reduction (%)')
plt.title('Defense Effectiveness')
plt.tight_layout()
plt.show()
```
