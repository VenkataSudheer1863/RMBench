# Results Visualization

Create publication-ready visualizations.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12
```

## Load Results

```python
with open('../results/example_results.json') as f:
    results = json.load(f)

print(f"Benchmark: {results['benchmark_name']}")
print(f"Timestamp: {results['timestamp']}")
```

## Comprehensive Results Table

```python
# Create summary table
summary_data = []
for model in results['models_evaluated']:
    summary_data.append({
        'Model': model['model_name'],
        'ASR ↓': model['metrics']['asr'],
        'Truth ↑': model['metrics']['truthfulness'],
        'Safety ↑': model['metrics']['tool_safety'],
        'Goal ↑': model['metrics']['goal_preservation']
    })

summary_df = pd.DataFrame(summary_data)
display(summary_df)

# Save for publication
summary_df.to_csv('../results/summary_table.csv', index=False)
print("\nTable saved to results/summary_table.csv")
```
