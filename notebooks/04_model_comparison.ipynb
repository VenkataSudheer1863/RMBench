# Model Comparison

Compare robustness across different models.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Multi-Metric Comparison

```python
import json

with open('../results/example_results.json') as f:
    results = json.load(f)

# Create comparison dataframe
data = []
for model in results['models_evaluated']:
    row = {'model': model['model_name']}
    row.update(model['metrics'])
    data.append(row)

df = pd.DataFrame(data)
display(df)
```

## Radar Chart

```python
# Radar chart showing model strengths
import numpy as np

metrics = ['truthfulness', 'tool_safety', 'goal_preservation', 'memory_integrity']
model_name = df['model'].iloc[0]
values = [df[m].iloc[0] for m in metrics]

angles = np.linspace(0, 2 * np.pi, len(metrics), endpoint=False).tolist()
values += values[:1]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(projection='polar'))
ax.plot(angles, values)
ax.fill(angles, values, alpha=0.25)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(metrics)
ax.set_title(f'Model: {model_name}')
plt.show()
```
