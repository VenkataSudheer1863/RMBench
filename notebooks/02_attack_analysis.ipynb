# Attack Analysis

Detailed analysis of different attack types and their effectiveness.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from rmbench.models import ModelRegistry
from rmbench.attacks import *
from rmbench.config import ModelBackend

sns.set_style('whitegrid')
```

## Load Results

Load attack evaluation results.

```python
# Load example results
import json

with open('../results/example_results.json') as f:
    results = json.load(f)

print(f"Models evaluated: {len(results['models_evaluated'])}")
```

## Attack Success Rates by Model

```python
# Extract ASR by model
models = [m['model_name'] for m in results['models_evaluated']]
asrs = [m['metrics']['asr'] for m in results['models_evaluated']]

plt.figure(figsize=(10, 6))
plt.barh(models, asrs, color='coral')
plt.xlabel('Attack Success Rate (%)')
plt.title('ASR by Model (Lower is Better)')
plt.tight_layout()
plt.show()
```

## Attack Type Effectiveness

```python
# Analyze attack types
model_data = results['models_evaluated'][0]
attack_types = list(model_data['attack_results'].keys())
success_rates = [model_data['attack_results'][a]['success_rate'] for a in attack_types]

plt.figure(figsize=(12, 6))
plt.bar(attack_types, success_rates, color='steelblue')
plt.xticks(rotation=45, ha='right')
plt.ylabel('Success Rate (%)')
plt.title('Attack Effectiveness by Type')
plt.tight_layout()
plt.show()
```
